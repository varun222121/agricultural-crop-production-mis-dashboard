```python
import numpy as np

# Ensure numeric fields are numeric
df["Area"] = pd.to_numeric(df["Area"], errors="coerce")
df["Production"] = pd.to_numeric(df["Production"], errors="coerce")
df["Yield"] = pd.to_numeric(df["Yield"], errors="coerce")

# Clean Year and create start_year
df["Year"] = df["Year"].astype("string").str.strip()

df["start_year"] = (
    df["Year"]
    .str.split("-")
    .str[0]
    .astype("Int64")
)

# Create a clean analysis dataset:
# keep records with area and production available and valid
analysis_df = df[
    (df["Area"].notna()) &
    (df["Production"].notna()) &
    (df["Area"] > 0) &
    (df["Production"] >= 0)
].copy()

# Recalculate yield from Area and Production
analysis_df["calculated_yield"] = (
    analysis_df["Production"] / analysis_df["Area"]
)
```

```python
total_production = analysis_df["Production"].sum()
total_area = analysis_df["Area"].sum()
overall_yield = total_production / total_area

number_of_states = analysis_df["State Name"].nunique()
number_of_districts = analysis_df["District Name"].nunique()
number_of_crops = analysis_df["Crop Name"].nunique()

latest_crop_year = (
    analysis_df.loc[
        analysis_df["start_year"].idxmax(),
        "Year"
    ]
)

print("Total Production:", total_production)
print("Total Area:", total_area)
print("Overall Yield:", overall_yield)
print("States:", number_of_states)
print("Districts:", number_of_districts)
print("Crops:", number_of_crops)
print("Latest Crop Year:", latest_crop_year)
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Production (Tonnes)",
        "Total Cultivated Area (Hectares)",
        "Overall Yield (Tonnes/Hectare)",
        "Number of States",
        "Number of Districts",
        "Number of Crops",
        "Latest Crop Year"
    ],
    "Value": [
        round(total_production, 2),
        round(total_area, 2),
        round(overall_yield, 3),
        number_of_states,
        number_of_districts,
        number_of_crops,
        latest_crop_year
    ]
})

kpi_summary
```

```python
# Production trend by crop year
year_summary = (
    analysis_df.groupby(["start_year", "Year"], as_index=False)
    .agg(
        total_production=("Production", "sum"),
        total_area=("Area", "sum")
    )
    .sort_values("start_year")
)

year_summary["weighted_yield"] = (
    year_summary["total_production"] /
    year_summary["total_area"]
)

# Production by crop type
crop_type_summary = (
    analysis_df.groupby("Crop Type", as_index=False)
    .agg(
        total_production=("Production", "sum"),
        total_area=("Area", "sum")
    )
)

crop_type_summary["weighted_yield"] = (
    crop_type_summary["total_production"] /
    crop_type_summary["total_area"]
)

crop_type_summary = crop_type_summary.sort_values(
    "total_production",
    ascending=False
)

# Top 10 states by production
state_summary = (
    analysis_df.groupby("State Name", as_index=False)
    .agg(
        total_production=("Production", "sum"),
        total_area=("Area", "sum"),
        district_count=("District Name", "nunique")
    )
)

state_summary["weighted_yield"] = (
    state_summary["total_production"] /
    state_summary["total_area"]
)

state_summary = state_summary.sort_values(
    "total_production",
    ascending=False
)

top_10_states = state_summary.head(10)

# Top 10 crops by production
crop_summary = (
    analysis_df.groupby(["Crop Name", "Crop Type"], as_index=False)
    .agg(
        total_production=("Production", "sum"),
        total_area=("Area", "sum")
    )
)

crop_summary["weighted_yield"] = (
    crop_summary["total_production"] /
    crop_summary["total_area"]
)

crop_summary = crop_summary.sort_values(
    "total_production",
    ascending=False
)

top_10_crops = crop_summary.head(10)
```

```python
# Top 10 crops by yield
top_10_yield_crops = (
    crop_summary[
        crop_summary["total_area"] >= 100
    ]
    .sort_values("weighted_yield", ascending=False)
    .head(10)
)

# Top 10 crops by cultivated area
top_10_area_crops = (
    crop_summary
    .sort_values("total_area", ascending=False)
    .head(10)
)

# Add production rank for a detailed crop table
crop_performance_table = crop_summary.copy()

crop_performance_table["production_rank"] = (
    crop_performance_table["total_production"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

crop_performance_table = crop_performance_table.sort_values(
    "production_rank"
)
```

```python
# State ranking by yield
state_yield_ranking = (
    state_summary[
        state_summary["total_area"] >= 1000
    ]
    .sort_values("weighted_yield", ascending=False)
)

# District-level performance
district_summary = (
    analysis_df.groupby(
        ["State Name", "District Name"],
        as_index=False
    )
    .agg(
        total_production=("Production", "sum"),
        total_area=("Area", "sum"),
        crops_count=("Crop Name", "nunique")
    )
)

district_summary["weighted_yield"] = (
    district_summary["total_production"] /
    district_summary["total_area"]
)

district_summary["production_rank_in_state"] = (
    district_summary.groupby("State Name")["total_production"]
    .rank(method="dense", ascending=False)
    .astype(int)
)

district_summary = district_summary.sort_values(
    ["State Name", "production_rank_in_state"]
)
```

```python
season_summary = (
    analysis_df.groupby("Season", as_index=False)
    .agg(
        total_production=("Production", "sum"),
        total_area=("Area", "sum")
    )
)

season_summary["weighted_yield"] = (
    season_summary["total_production"] /
    season_summary["total_area"]
)

season_summary = season_summary.sort_values(
    "total_production",
    ascending=False
)

season_crop_type_matrix = (
    analysis_df.groupby(
        ["Season", "Crop Type"],
        as_index=False
    )
    .agg(
        total_production=("Production", "sum"),
        total_area=("Area", "sum")
    )
)

season_crop_type_matrix["weighted_yield"] = (
    season_crop_type_matrix["total_production"] /
    season_crop_type_matrix["total_area"]
)
```

```python
data_quality_summary = pd.DataFrame({
    "Data Quality Check": [
        "Total Records",
        "Missing Area",
        "Missing Production",
        "Missing Yield",
        "Zero Production",
        "Zero Yield",
        "Duplicate Rows",
        "Invalid Crop Year",
        "Non-Tonnes Production Units",
        "Non-Hectare Area Units"
    ],
    "Count": [
        len(df),
        df["Area"].isna().sum(),
        df["Production"].isna().sum(),
        df["Yield"].isna().sum(),
        (df["Production"] == 0).sum(),
        (df["Yield"] == 0).sum(),
        df.duplicated().sum(),
        (~df["Year"].str.match(r"^\d{4}-\d{4}$", na=False)).sum(),
        (df["Production Unit"] != "Tonnes").sum(),
        (df["Area Unit"] != "Hectare").sum()
    ]
})

data_quality_summary
```
